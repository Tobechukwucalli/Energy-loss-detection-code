```python
import pandas as pd

def standardize_house_type(house):

    if pd.isna(house):
        return "Unknown"

    house = str(house).strip().lower()

    if "cold" in house:
        return "Cold Room"

    elif "3 bedroom" in house or "3 bedrooom" in house:
        return "3 Bedroom Flat"

    elif "2 bedroom" in house:
        return "2 Bedroom Flat"

    elif "market shop" in house:
        return "Shop"

    elif house == "shop":
        return "Shop"

    elif "store" in house:
        return "Store Room"

    elif "residential" in house:
        return "Residential"

    else:
        return house.title()

energy_theft_data_info["House_Type_Clean"] = (
    energy_theft_data_info["House Type"]
    .apply(standardize_house_type)
)
```

```python
energy_theft_data_info["House_Type_Clean"].value_counts()
```

```python
expected_load = {

    "Cold Room":35,

    "3 Bedroom Flat":20,

    "2 Bedroom Flat":10,

    "Residential":10,

    "Shop":15,

    "Store Room":15,

    "Unknown":10

}
```

```python
energy_theft_data_info["Expected_Current_Load"] = (
    energy_theft_data_info["House_Type_Clean"]
    .map(expected_load)
)
```

```python
energy_theft_data_info[
[
"House_Type_Clean",
"Expected_Current_Load"
]
].head(10)
```

```python
energy_theft_data_info[
[
"House_Type_Clean",
"Expected_Current_Load"
]
].head(10)
```

```python
energy_theft_data_info["Load_Deficit"] = (

    energy_theft_data_info["Expected_Current_Load"]

    -

    energy_theft_data_info["Current load"]

)
```

```python
energy_theft_data_info["ColdRoomFlag"] = (

    (energy_theft_data_info["House_Type_Clean"]=="Cold Room")

    &

    (energy_theft_data_info["Current load"]<35)

)
```

```python
energy_theft_data_info["ColdRoomFlag"].sum()
```

```python
energy_theft_data_info["Feeder33kVFlag"] = (

    energy_theft_data_info["Feeder"]

    .str.contains("33",case=False,na=False)

)

energy_theft_data_info["LowLoad33kV"] = (

    energy_theft_data_info["Feeder33kVFlag"]

    &

    (energy_theft_data_info["Current load"]<15)

)
```

```python
energy_theft_data_info["LowLoad33kV"].sum()
```

```python
energy_theft_data_info["Rule_Based_Suspicion"] = (

    energy_theft_data_info["ColdRoomFlag"]

    |

    energy_theft_data_info["LowLoad33kV"]

    |

    (energy_theft_data_info["Load_Deficit"]>5)

)
```

```python
def calculate_risk(row):

    score = 0

    if row["ColdRoomFlag"]:
        score += 40

    if row["LowLoad33kV"]:
        score += 35

    if row["Load_Deficit"] > 5:
        score += 25

    return min(score,100)

energy_theft_data_info["Risk_Score"] = (
    energy_theft_data_info.apply(calculate_risk,axis=1)
)
```

```python
def risk_level(score):

    if score >= 70:
        return "High Risk"

    elif score >= 40:
        return "Medium Risk"

    else:
        return "Low Risk"

energy_theft_data_info["Risk_Level"] = (
    energy_theft_data_info["Risk_Score"]
    .apply(risk_level)
)
```

```python
energy_theft_data_info[
[
"Customer Account Number",
"House_Type_Clean",
"Current load",
"Expected_Current_Load",
"Load_Deficit",
"Risk_Score",
"Risk_Level"
]
].sort_values(
by="Risk_Score",
ascending=False
)
```

```python
energy_theft_data_info["Theft"] = energy_theft_data_info["Remark"].map({
    "Not bypassing": 0,
    "Caught on bypass": 1
})
```

```python
from sklearn.preprocessing import LabelEncoder

house_encoder = LabelEncoder()
feeder_encoder = LabelEncoder()
tariff_encoder = LabelEncoder()

energy_theft_data_info["House_Type_Encoded"] = house_encoder.fit_transform(
    energy_theft_data_info["House_Type_Clean"]
)

energy_theft_data_info["Feeder_Encoded"] = feeder_encoder.fit_transform(
    energy_theft_data_info["Feeder"]
)

energy_theft_data_info["Tariff_Encoded"] = tariff_encoder.fit_transform(
    energy_theft_data_info["Tariff Band"]
)
```

```python
features = [

    "Current load",

    "Expected_Current_Load",

    "Load_Deficit",

    "Risk_Score",

    "House_Type_Encoded",

    "Feeder_Encoded",

    "Tariff_Encoded"

]
```

```python
X = energy_theft_data_info[features]

y = energy_theft_data_info["Theft"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(

    n_estimators=200,

    random_state=42

)

rf_model.fit(X_train, y_train)
```

```python
predictions = rf_model.predict(X_test)

probabilities = rf_model.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

print("Accuracy:", accuracy_score(y_test, predictions))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, predictions))

print("\nClassification Report\n")
print(classification_report(y_test, predictions))
```

```python
import pandas as pd

importance = pd.DataFrame({

    "Feature": features,

    "Importance": rf_model.feature_importances_

})

importance = importance.sort_values(

    by="Importance",

    ascending=False

)

print(importance)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

comparison = pd.DataFrame(columns=[
    "Model",
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
])

models = {
    "Random Forest": rf,
    "Decision Tree": dt,
    "Logistic Regression": lr,
    "SVM": svm,
    "XGBoost": xgb
}

for name, model in models.items():

    prediction = model.predict(X_test)

    comparison.loc[len(comparison)] = [

        name,

        accuracy_score(y_test, prediction),

        precision_score(y_test, prediction),

        recall_score(y_test, prediction),

        f1_score(y_test, prediction)

    ]

comparison
```
